```python
import csv
data = [ ['name', 'age', 'place'],
        ['ali', '26', 'faisalabad'],
         ['mudasir', '22', 'lahore'],
         ['abbas', '29', 'ryk'],
         ['abid', '33', 'dgk'] ]
with open('people.csv', 'w', newline='') as file:
  writer = csv.writer(file)
  writer.writerows(data)
print('new csv file!')
```

```python
import csv
with open('people.csv', 'r') as file:
  reader = csv.reader(file)
  for row in reader:
    print('row', row)
```

```python
import csv
with open('people.csv', 'r') as file:
  read = csv.DictReader(file)
  for row in read:
      print(row['name'], '-', row['age'], '-', row['place'])
```

```python
new_data = ['saima', '25', 'lhr']
with open('people.csv', 'a') as file:
  writer = csv.writer(file)
  writer.writerow(new_data)
print('updated file')
```
